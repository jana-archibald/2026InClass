## Plotting Normal Distributions

You are not required to use this functio, but sometimes it can be convinient to be able to plot (and shade) normal distributions.

```r
normalplot<-function(m,sd,region=0){
  x<-seq(m-(3.5)*sd,m+(3.5)*sd,length=1000)
  y<-dnorm(x,m,sd)
  plot(x,y,type="l",xlab="",ylab="", bty="n", yaxt="n")
  z<-x[x>region[1]]
  z<-z[z<region[2]]
  polygon(c(region[1],z,region[2]),
          c(0,dnorm(z,m,sd),0),col="gray")
  abline(v=m)
  abline(h=0)}
```

```r
normalplot(100,15)
```

## Scores on an IQ test are approximately normally distributed with a mean of 100 and a standard deviation of 15.

## What is the probability that a person selected at random has ...

### ... an IQ less than 75

```r
normalplot(100,15,c(0,75))
```

```r
pnorm(75,100,15)
```

$$P(X<75)=pnorm(75,100,15)=0.04779...$$

### ... an IQ above  96

```r
pnorm(96,100,15)
```

```r
1-pnorm(96,100,15)
```

```r
pnorm(96,100,15,lower.tail=F)
```

```r
normalplot(100,15,c(96,1000))
```

(c)  $$P(X>96)=0.605137089535975$$

### ... an IQ between 105 and 135

```r
pnorm(135,100,15)
pnorm(105,100,15)
pnorm(135,100,15)- pnorm(105,100,15)
```

```r
normalplot(100,15,c(105,135))
```

(a)  $$P(105<X<135)=P(X<135)-P(X<105)=0.990184671371355-0.630558659818236=0.359626011553118$$

### ...  an IQ between  90 and 120

```r
normalplot(100,15,c(90,120))
```

```r
pnorm(120,100,15)-pnorm(90,100,15)
```

(d)  $$P(90<X<120)=P(X<120)-P(X<90)=0.9088-0.2525=0/6563$$
